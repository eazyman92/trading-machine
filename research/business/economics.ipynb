# Trading app economics: illustrative scenarios

No observed customers, billing, retention, or capacity measurements exist. USD scenarios dated 3 November 2026; months are relative to commercial launch, not today's calendar. This is a table-only calculation companion to the chat answer, not a business forecast.

## Context and assumptions

A $39 weighted monthly subscription; 60 AI calls per paying user with 5,000 input and 2,000 billable output tokens each; three limited active-free users subsidized per payer; market-data allowance of $3/payer without a commercial quote. All workload, customer growth, pricing, churn, acquisition, staffing and capacity inputs are planning assumptions. Published vendor prices anchor only some inputs. Model excludes tax, upfront development/security/legal launch work, unquoted mandatory regulatory/commercial charges, and customers' broker trading costs. Gross MRR is not profit or cash collected.

```python
import json
from pathlib import Path
from model import load, build
assumptions = load()
results = build(assumptions)
print(json.dumps({'average_price': results['average_monthly_price'], 'variable_components': results['variable_components']}, indent=2))
```

## Scale economics

Direct service costs include the modeled free-user subsidy, support, fees/refunds, and shared operating allowance. Operating balance also subtracts engineering/admin and acquisition allowances. Scale acquisition budget assumes gross monthly additions equal to 10% of opening payer count at $30 each; with 5% churn this approximates 5% net monthly growth, not a measured funnel.

```python
for row in results['scale']:
    print(f"{row['paying_users']:,} payers | gross MRR ${row['gross_monthly_subscription_revenue']:,.0f} | direct ${row['direct_service_cost']:,.0f} | direct/payer ${row['direct_cost_per_payer']:.2f} | operating balance ${row['modeled_operating_balance_before_tax']:,.0f}")
```

## Growth scenarios

Paying users = prior paying users × 95% + gross new paying users. Gross acquisition schedules are explicit assumptions in assumptions.json. Expected counts may be fractional and are rounded for display. Revenue shown is end-month gross subscription run rate. Any modeled monthly/cumulative balances assume end-month payers pay a full month and ignore invoice timing/proration; they are not a cash-flow forecast.

```python
for row in results['growth']:
    if row['month_after_commercial_launch'] in [3, 6, 12]:
        print(f"{row['scenario']} month {row['month_after_commercial_launch']}: approximately {row['paying_users']:,.0f} payers, ${row['gross_monthly_subscription_revenue']:,.0f} gross MRR")
```

## Sensitivity and verification

The sensitivity cases are alternative assumptions, not confidence intervals. Model assertions independently check the $39 price mix, token-cost calculation, a 1,000-user cost stack, growth recurrence and revenue/cost reconciliation. Breakeven holds 1,000-user fixed capacity constant and must not be treated as a universal threshold.

```python
for row in results['sensitivity_at_1000_payers']:
    print(f"{row['case']}: operating balance ${row['modeled_operating_balance_before_tax']:,.0f}/month")
print('All embedded arithmetic checks passed.')
```

## Takeaways

Under the base assumptions, 1,000 payers produce $39,000 gross MRR, $12,450 direct service cost, and $13,550 operating balance after the assumed $10,000 engineering/admin and $3,000 acquisition allowances. Raising data allowance to $25/payer makes that balance negative.
